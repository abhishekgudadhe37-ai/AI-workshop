```python
import numpy as np  
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data= {
    "name": ["Abhishek", "Rohit", "Sakshi", "Anjali", "Ramesh"],
    "maths": [85, 90, 78, 92, 88],
    "science": [92, 88, 95, 90, 85],
    "english": [78, 85, 88, 90, 92] 
}
df = pd.DataFrame(data)
print(df)
```

```python
plt. figure()
plt. bar(df["name"], df["maths"], color='b', label='Maths')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Subject-wise Performance')
plt.show()
```

```python
#average marks per subject
avg_maths = df["maths"].mean()
avg_science = df["science"].mean()
avg_english = df["english"].mean()

print(f"Average Marks - Maths: {avg_maths}, Science: {avg_science}, English: {avg_english}")
```

```python
#import lib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
## Load IOT data
df = pd.read_csv('iot_sensor_data.csv')
df{"Timestamp"} = pd.to_datetime(df["Timestamp"])
print("iot_sensor_data\n",df)
```

```python
#statical analysis 
temp = df["Temperature"].values
mean_temp = np.mean(temp)#mean (avg)
std_temp = np.std(temp)#standard deviation
print("\n Mean Temperature:", mean_temp)
print("Standard Deviation of Temperature:", std_temp)
```
